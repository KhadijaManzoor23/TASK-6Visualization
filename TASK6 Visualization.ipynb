```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))
df['Nationality'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Nationalities')
plt.xlabel('Country')
plt.ylabel('Number of Players')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
df.groupby('Year')['Overall'].mean().plot(kind='line')
plt.title('Average Player Rating Over the Years')
plt.xlabel('Year')
plt.ylabel('Average Rating')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 8))
df['Position'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Players by Position')
plt.ylabel('')
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.pairplot(df[['Age', 'Overall', 'Potential']])
plt.suptitle('Pairplot: Age, Overall, Potential', y=1.02)
plt.show()
```
